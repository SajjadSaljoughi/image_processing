```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
def draw_half_circle_rounded(image,radius,color):
    height, width = image.shape[0:2]
    center = (width // 2, height - 25)
    axes = (radius, radius)
    angle = 0
    startAngle = 180
    endAngle = 360
    thickness = 20
    cv2.ellipse(image,center,axes,angle,startAngle,endAngle,color,thickness)
```

```python
red_layout = np.ones((400,800),dtype=np.uint8) * 255
green_layout = np.ones((400,800),dtype=np.uint8) * 255
blue_layout = np.ones((400,800),dtype=np.uint8) * 255
```

```python
draw_half_circle_rounded(red_layout,350,255)
draw_half_circle_rounded(green_layout,350,0)
draw_half_circle_rounded(blue_layout,350,0)

draw_half_circle_rounded(red_layout,330,255)
draw_half_circle_rounded(green_layout,330,165)
draw_half_circle_rounded(blue_layout,330,0)

draw_half_circle_rounded(red_layout,310,255)
draw_half_circle_rounded(green_layout,310,255)
draw_half_circle_rounded(blue_layout,310,0)

draw_half_circle_rounded(red_layout,290,0)
draw_half_circle_rounded(green_layout,290,200)
draw_half_circle_rounded(blue_layout,290,0)

draw_half_circle_rounded(red_layout,270,0)
draw_half_circle_rounded(green_layout,270,255)
draw_half_circle_rounded(blue_layout,270,255)

draw_half_circle_rounded(red_layout,250,0)
draw_half_circle_rounded(green_layout,250,0)
draw_half_circle_rounded(blue_layout,250,255)

draw_half_circle_rounded(red_layout,230,255)
draw_half_circle_rounded(green_layout,230,0)
draw_half_circle_rounded(blue_layout,230,255)
```

```python
image = cv2.merge([red_layout,green_layout,blue_layout])
plt.axis('off')
plt.imshow(image)
plt.show()
```

```python
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
cv2.imwrite("output/result.jpg",image)
```

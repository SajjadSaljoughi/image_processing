```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("input/image.jpg")
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.imshow(image)
plt.show()
```

```python
rows,cols = image.shape[:2]
#Help
#image[i,j,0] = red
#image[i,j,1] = green
#image[i,j,2] = blue
for i in range(rows):
    for j in range(cols):
        if image[i,j,0] < 50 and image[i,j,1] > 100 and image[i,j,2] > 100:
            image[i,j,0] = 255
            image[i,j,1] = 0
            image[i,j,2] = 0
        elif image[i,j,0] > 100 and image[i,j,1] < 50 and image[i,j,2] > 100:
            image[i,j,0] = 0
            image[i,j,1] = 255
            image[i,j,2] = 0
        elif image[i,j,0] > 100 and image[i,j,1] > 100 and image[i,j,2] < 50:
            image[i,j,0] = 0
            image[i,j,1] = 0
            image[i,j,2] = 255
```

```python
plt.imshow(image)
plt.show()
```

```python
result = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
cv2.imwrite("output/result.jpg",result)
```

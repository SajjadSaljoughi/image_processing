```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("input/test.jpg")
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.imshow(image)
plt.show()
```

```python
def color_to_gray(red, green, blue):
    return red // 3 + green // 3 + blue // 3
```

```python
r,g,b = cv2.split(image)
gray = color_to_gray(r, g, b)
plt.imshow(gray, cmap="gray")
plt.show()
plt.show()
```

```python
cv2.imwrite("output/result.jpg",gray)
```

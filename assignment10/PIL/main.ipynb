```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image,ImageDraw,ImageFont,ImageText
import arabic_reshaper
from bidi.algorithm import get_display
```

```python
image = Image.open("input/image.jpg")
plt.imshow(image)
plt.show()
```

```python
def add_font(background):
    rgb = cv2.cvtColor(background, cv2.COLOR_BGR2RGB)
    _image  = Image.fromarray(rgb)
    draw = ImageDraw.Draw(_image)
    font = ImageFont.truetype("input/tahoma.ttf", 65)
    text = "سجاد سلجوقی"
    reshaped_text = arabic_reshaper.reshape(text)
    bidi_text = get_display(reshaped_text)
    draw.text((100, 0), bidi_text , font=font, fill=(0, 0, 0))
    _image = cv2.cvtColor(np.array(_image), cv2.COLOR_RGB2BGR)
    return _image
```

```python
image_cv2 = np.array(image)
result = add_font(image_cv2)
result = cv2.cvtColor(result, cv2.COLOR_BGR2RGB)
cv2.imwrite("output/result.jpg", result)
```

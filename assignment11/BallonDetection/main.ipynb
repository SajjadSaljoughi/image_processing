```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('input/image.jpg')
image_hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
h,s,v = cv2.split(image_hsv)
mask = ((h > 5) & (h < 175)) | (s == 0)
v[mask] = 0
result = cv2.merge([h,s,v])
_result = cv2.cvtColor(result, cv2.COLOR_HSV2BGR)
plt.imshow(_result)
plt.show()
cv2.imwrite('output/ballon.png', _result)
```

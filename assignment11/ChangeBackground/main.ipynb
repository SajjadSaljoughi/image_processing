```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('input/image.jpg')
image_hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
sky = cv2.imread("input/sky.jpg")
sky = cv2.resize(sky, (image.shape[1], image.shape[0]))
sky_hsv = cv2.cvtColor(sky, cv2.COLOR_BGR2HSV)
h,s,v = cv2.split(image_hsv)
mask = ((h > 35) & (h < 85)) & (s > 75)
image_hsv[mask] = sky_hsv[mask]
_result = cv2.cvtColor(image_hsv, cv2.COLOR_HSV2BGR)
r = cv2.cvtColor(image_hsv, cv2.COLOR_HSV2RGB)
plt.imshow(r)
plt.show()
cv2.imwrite('output/result.png', _result)
```
